```python
import numpy as np

# Fill missing numerical values with median for trader_df
trader_df.fillna(trader_df.median(numeric_only=True), inplace=True)

# Fill missing numerical values with median for sentiment_df
sentiment_df.fillna(sentiment_df.median(numeric_only=True), inplace=True)
```

```python
# Fill missing categorical values with mode for trader_df
for col in trader_df.select_dtypes(include='object'):
    trader_df[col].fillna(trader_df[col].mode()[0], inplace=True)

# Fill missing categorical values with mode for sentiment_df
for col in sentiment_df.select_dtypes(include='object'):
    sentiment_df[col].fillna(sentiment_df[col].mode()[0], inplace=True)
```

```python
if 'date' in sentiment_df.columns:
    sentiment_df['date'] = pd.to_datetime(sentiment_df['date'])
```

Feature Selection

```python
import pandas as pd
import numpy as np

# Applying transformations to sentiment_df
# 1. date -> trading_date
# Converts a string or standard date into a proper datetime object
sentiment_df['trading_date'] = pd.to_datetime(sentiment_df['date'])

# 6. market_sentiment -> fear / greed
# Assuming 'value' column in sentiment_df represents a sentiment score (e.g., 0 to 100)
# Here, anything below 50 becomes 'Fear', and 50 or above becomes 'Greed'.
sentiment_df['market_sentiment'] = np.where(sentiment_df['value'] < 50, 'Fear', 'Greed')


# Applying transformations to trader_df
# 2. trading_direction -> Buy/Sell direction
# Assuming 'Direction' column in trader_df uses something like 1/-1, or 'B'/'S'.
# You can map it to explicit 'Buy' and 'Sell' labels (or vice versa for ML).
direction_mapping = {1: 'Buy', -1: 'Sell', 'B': 'Buy', 'S': 'Sell'}
trader_df['Buy/Sell_direction'] = trader_df['Direction'].map(direction_mapping)
# Note: If feeding this into a model, you'd actually want 1 and 0 instead of text.

# 3. profit_tag -> profit/loss
# Assuming 'Closed PnL' in trader_df is a numeric value where > 0 is profit and <= 0 is loss
trader_df['profit/loss'] = np.where(trader_df['Closed PnL'] > 0, 'Profit', 'Loss')

# 4. trade_size -> absolute_trade_size
# Calculates the absolute value (removes any negative signs) from 'Size Tokens'
trader_df['absolute_trade_size'] = trader_df['Size Tokens'].abs()


# 5. trade_value -> Execution price * trade_size
# Multiplies the 'Execution Price' by the 'Size Tokens' to get the total value of the trade.
# Note: Ensure both columns are numeric before multiplying.
trader_df['trade_value'] = trader_df['Execution Price'] * trader_df['Size Tokens']
```

```python
# Define features for sentiment_df
sentiment_features = ['value', 'classification', 'trading_date', 'market_sentiment']
sentiment_df = sentiment_df[sentiment_features]

# Define features for trader_df
trader_features = ['Execution Price', 'Size Tokens', 'Size USD', 'Side', 'Direction', 'Closed PnL', 'Fee', 'Buy/Sell_direction', 'profit/loss', 'absolute_trade_size', 'trade_value']
trader_df = trader_df[trader_features]

print("Features selected for sentiment_df and trader_df.")
```

```python
# Outlier Handling for trader_df
print("Applying outlier handling to trader_df...")
initial_trader_rows = trader_df.shape[0]
for col in trader_df.select_dtypes(include=np.number).columns:
    q1 = trader_df[col].quantile(0.25)
    q3 = trader_df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    trader_df = trader_df[(trader_df[col] >= lower) & (trader_df[col] <= upper)]
print(f"Removed {initial_trader_rows - trader_df.shape[0]} outliers from trader_df. New shape: {trader_df.shape}")

# Outlier Handling for sentiment_df
print("\nApplying outlier handling to sentiment_df...")
initial_sentiment_rows = sentiment_df.shape[0]
for col in sentiment_df.select_dtypes(include=np.number).columns:
    q1 = sentiment_df[col].quantile(0.25)
    q3 = sentiment_df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    sentiment_df = sentiment_df[(sentiment_df[col] >= lower) & (sentiment_df[col] <= upper)]
print(f"Removed {initial_sentiment_rows - sentiment_df.shape[0]} outliers from sentiment_df. New shape: {sentiment_df.shape}")
```
